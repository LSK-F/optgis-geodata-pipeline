### Why use H3 Resolution 8?

In the context of transport planning and urban infrastructure analysis, this resolution offers the ideal balance:

* Each hexagon covers approximately 0.73 km², representing a scale broadly consistent with short urban trips and walkable areas, although actual walking times depend on local conditions.
* A uniform grid reduces the influence of administrative boundaries and improves the comparability of density measures across the Metropolitan Region, mitigating part of the Modifiable Areal Unit Problem (MAUP).
* The resolution is detailed enough to identify local demand patterns while providing spatial aggregation that helps avoid exposing the exact location of individual observations.
